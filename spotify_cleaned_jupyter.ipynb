```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
engine = create_engine(
    "postgresql+psycopg2://postgres:SJNCTREWAmean786@localhost:5432/spotify_analisis"
) 
```

```python
# data import
query = "SELECT * FROM spotify"
df = pd.read_sql(query, engine)
print("✅ Data Imported Successfully")
```

```python
# correlation
numeric_df = df.select_dtypes(include="number")
numeric_df.corr() 
```

```python
# popularity distribution
plt.figure(figsize=(8,5))
plt.hist(df["popularity"], bins=20)
plt.title("Popularity Distribution")
plt.xlabel("Popularity")
plt.ylabel("Frequency")
plt.show() 
```

```python
# duration distribution / tempo distribution
plt.figure(figsize=(8,5))
plt.hist(df["duration_ms"]/60000, bins=20)
plt.title("Song Duration")
plt.xlabel("Minutes")
plt.ylabel("Songs")
plt.show() 
```

```python
# top 10 artist
top_artist = df["artist_name"].value_counts().head(10)
top_artist.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 Artists")
plt.xlabel("Artist")
plt.ylabel("Songs")
plt.xticks(rotation=45)
plt.show()  
```

```python
# top 10 genres
top_genre = df["genre"].value_counts().head(10)
top_genre.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 Genres")
plt.xlabel("Genre")
plt.ylabel("Songs")
plt.xticks(rotation=45)
plt.show() 
```

```python
# top 10 countries
top_country = df["country"].value_counts().head(10)
top_country.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 Countries")
plt.xlabel("Country")
plt.ylabel("Songs")
plt.xticks(rotation=45)
plt.show() 
```

```python
# danceability vs popularity
plt.figure(figsize=(8,6))
plt.scatter(df["danceability"], df["popularity"])
plt.xlabel("Danceability")
plt.ylabel("Popularity")
plt.title("Danceability vs Popularity")
plt.show() 
```

```python
# box plot
plt.figure(figsize=(8,5))
plt.boxplot(df["popularity"])
plt.title("Popularity Outliers")
plt.show() 
```

```python
# correlation heatmap
corr = numeric_df.corr()
plt.figure(figsize=(12,8))
plt.imshow(corr)
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.colorbar()
plt.title("Correlation Heatmap")
plt.show()
```

```python
# top artists
artist_dashboard=df.groupby("artist_name").agg({
    "stream_count":"sum",
    "popularity":"mean"
}).sort_values("stream_count",ascending=False)

artist_dashboard.head()
```

```python
# country dashboard
country_dashboard=df.groupby("country").agg({
    "stream_count":"sum",
    "popularity":"mean"
})

country_dashboard
```
